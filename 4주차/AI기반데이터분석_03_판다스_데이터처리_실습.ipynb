## 1. 파이썬 파일 처리

```python
sample_file = 'data/sample.txt'
```

```python
# 파일을 open하여 내용 출력
with open(sample_file, mode='rt') as f:
    contents = f.read()
    print(contents)
```

```python
# 파일을 open하고 한 줄씩 출력하기
with open(sample_file, mode="r") as f:
    lines = f.readlines()
    print(lines)
    for line in lines:
     print(line, end="")
```

```python
# 파일객체에서 한 줄씩 읽어 오기
# 파일 객체 f를 반복문에서 사용하면 파일을 한 줄씩 읽으며, 각 줄 끝의 \n도 포함해서 가져옵
with open (sample_file, mode="r") as f:
    for line in f:
       print(line, end="")
```

```python
# 여러 개의 파일 읽기
import glob

files = glob.glob("data/ *. txt")
print(files)
for file in files:
    with open(file, mode="r") as f:
        print(f.read())
    print(" ---- 끝 ---- ")
```
