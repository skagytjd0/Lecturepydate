## 1. 파이썬 파일 처리

```python
sample_file = 'C:\\Users\\1\\Desktop\\강의자료\\2학기\\빅데이터\\4주차\\data\\sample.txt'
```

```python
# 파일을 open하여 전체 내용을 읽어서 출력
with open(sample_file, mode='rt') as f:
    content = f.read()
    print(content)
print("----- 끝 -----")
```

```python
# 파일을 open하고 한 줄씩 읽어서 출력하기 : readlines(), readline()
with open(sample_file, 'r') as f:
    lines = f.readlines()
    print(lines)


print("----- 끝 -----")
```

```python
# 파일을 open하고 한 줄씩 읽어서 출력하기 : readlines(), readline()
with open(sample_file, 'r') as f:
    line = f.readline()
    while line:
        print(line, end="")
        line = f. readline()
```

```python
# 파일객체에서 한 줄씩 읽어 오기
# 파일 객체 f를 반복문에서 사용하면 파일을 한 줄씩 읽으며,
# 각 줄 끝의 \n도 포함해서 가져옵
with open(sample_file, 'r') as f:
    for line in f:
        print(line, end="")

print("----- 끝 -----")
```

```python
# 여러 개의 파일 읽기 (모든 txt 파일을 읽어서 출력하기)
import glob
txt_file_pattern = "C:\\Users\\1\\Desktop\\강의자료\\2학기\\빅데이터\\4주차\\data\\*.txt"
files = glob.glob(txt_file_pattern)
print(files)
for file in files:
    print(file)
    with open(file, "r") as f:
        print(f.read())
print(" -----끝---- ")
```
